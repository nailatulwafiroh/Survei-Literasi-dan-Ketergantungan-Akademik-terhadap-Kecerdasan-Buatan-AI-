# 3.Populasi, Sampel, Margin of Error & Ukuran Sampling
<div style="background:#1E3A5F;padding:14px 20px;border-radius:10px;color:white;font-family:'Segoe UI',sans-serif;margin:12px 0 4px 0">
  <p style="margin:6px 0 0 0;font-size:13px;opacity:.8">Perhitungan statistik sampling menggunakan rumus Slovin dan Cochran dengan koreksi populasi terbatas (FPC).</p>
</div>

```python
# @title Populasi, Sampel, Margin of Error & Ukuran Sampling
# ── 1. Parameter & Kalkulasi ─────────────────────────────────────────────────
N_populasi = 2000
n_sampel   = len(df)
z          = 1.96
p          = 0.5
confidence = 0.95

MoE_raw = z * math.sqrt(p * (1 - p) / n_sampel)
fpc     = math.sqrt((N_populasi - n_sampel) / (N_populasi - 1))
MoE     = MoE_raw * fpc

n_slovin  = N_populasi / (1 + N_populasi * 0.10**2)
n0        = (z**2 * p * (1 - p)) / (0.10**2)
n_cochran = n0 / (1 + (n0 - 1) / N_populasi)

# ── 2. Komponen Visual HTML ──────────────────────────────
sampling_html = f"""
<link href="https://fonts.googleapis.com/css2?family=Plus+Jakarta+Sans:wght@400;500;600;700;800&display=swap" rel="stylesheet">
<style>
  .main-container-white {{
    background: #ffffff !important;
    padding: 30px;
    border-radius: 16px;
    border: 1px solid #e2e8f0;
    font-family: 'Plus Jakarta Sans', sans-serif;
    margin: 15px 0;
  }}
  .sm-title {{
    font-size: 20px;
    font-weight: 800;
    color: #0a1f44;
    margin-bottom: 20px;
    letter-spacing: -0.02em;
  }}
  .sm-grid {{
    display: grid;
    grid-template-columns: repeat(5, 1fr);
    gap: 12px;
    margin-bottom: 25px;
  }}
  .sm-item {{
    background: #f8fafc;
    padding: 14px;
    border-radius: 12px;
    border: 1px solid #edf2f7;
    box-shadow: 0 2px 8px rgba(0,0,0,0.01);
  }}
  .sm-label {{
    font-size: 9.5px;
    font-weight: 700;
    color: #64748b;
    text-transform: uppercase;
    letter-spacing: 0.05em;
    margin-bottom: 6px;
  }}
  .sm-value {{
    font-size: 16px;
    font-weight: 800;
    color: #1e3a8a;
    margin-bottom: 4px;
  }}
  .sm-desc {{
    font-size: 10px;
    color: #94a3b8;
    line-height: 1.3;
  }}
  .narasi-box {{
    background: #f0fdf4;
    border-left: 4px solid #22c55e;
    padding: 16px;
    border-radius: 0 12px 12px 0;
    margin-top: 25px;
    font-size: 13px;
    line-height: 1.6;
    color: #166534;
  }}
</style>

<div class="main-container-white">
  <div class="sm-title">Analisis Ukuran Sampling &amp; Representativitas Data</div>

  <div class="sm-grid">
    <div class="sm-item">
      <div class="sm-label">Populasi (N)</div>
      <div class="sm-value">{N_populasi:,}</div>
      <div class="sm-desc">Estimasi mahasiswa aktif ITSB</div>
    </div>
    <div class="sm-item">
      <div class="sm-label">Sampel (n)</div>
      <div class="sm-value">{n_sampel}</div>
      <div class="sm-desc">Jumlah responden valid</div>
    </div>
    <div class="sm-item">
      <div class="sm-label">MoE + FPC</div>
      <div class="sm-value">±{MoE*100:.2f}%</div>
      <div class="sm-desc">Tingkat kepercayaan 95%</div>
    </div>
    <div class="sm-item">
      <div class="sm-label">Min Slovin</div>
      <div class="sm-value">{n_slovin:.0f}</div>
      <div class="sm-desc">Batas toleransi e = 10%</div>
    </div>
    <div class="sm-item">
      <div class="sm-label">Min Cochran</div>
      <div class="sm-value">{n_cochran:.0f}</div>
      <div class="sm-desc">CI 95%, e = 10%</div>
    </div>
  </div>

  <div id="plotly-graph-area"></div>
"""

# ── 3. Visualisasi Plotly Interaktif ─────────────────────────────────────────
ns      = np.arange(10, 500, 5)
fpc_arr = np.sqrt((N_populasi - ns) / (N_populasi - 1))
moes    = 1.96 * np.sqrt(0.5 * 0.5 / ns) * fpc_arr * 100

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Kurva MoE vs Ukuran Sampel (CI 95%)", "Proporsi Sampel terhadap Populasi"),
    specs=[[{"type": "xy"}, {"type": "pie"}]],
    column_widths=[0.58, 0.42]
)

# Subplot 1: Line Chart Interaktif
fig.add_trace(go.Scatter(
    x=ns, y=moes, mode="lines", name="MoE + FPC",
    line=dict(color="#2E86C1", width=2.5),
    fill="tozeroy", fillcolor="rgba(46,134,193,0.05)",
    hovertemplate="Sampel: %{x}<br>MoE: %{y:.2f}%<extra></extra>"
), row=1, col=1)

fig.add_vline(
    x=n_sampel, line_dash="dash", line_color="#E74C3C", line_width=1.5,
    annotation_text=f"Sampel Aktual (n={n_sampel})",
    annotation_position="top right",
    annotation_font=dict(color="#E74C3C", size=9),
    row=1, col=1
)

fig.add_hline(
    y=MoE*100, line_dash="dot", line_color="#27AE60", line_width=1.5,
    annotation_text=f"MoE {MoE*100:.2f}%",
    annotation_position="right",
    annotation_font=dict(color="#27AE60", size=9),
    row=1, col=1
)

# Subplot 2: Donut Chart Interaktif
fig.add_trace(go.Pie(
    labels=[f"Sampel", f"Sisa Populasi"],
    values=[n_sampel, N_populasi - n_sampel],
    marker_colors=["#2E86C1", "#F1F5F9"],
    marker=dict(line=dict(color="#FFFFFF", width=2)),
    hole=0.5,
    textinfo="percent",
    hovertemplate="%{label}: %{value} mhs (%{percent})<extra></extra>"
), row=1, col=2)

fig.update_xaxes(title_text="Ukuran Sampel (n)", showgrid=True, gridcolor="#E2E8F0", row=1, col=1)
fig.update_yaxes(title_text="Margin of Error (%)", showgrid=True, gridcolor="#E2E8F0", row=1, col=1)

fig.update_layout(
    margin=dict(t=40, b=10, l=10, r=10),
    height=340,
    showlegend=True,
    legend=dict(orientation="h", yanchor="bottom", y=-0.25, xanchor="center", x=0.5),
    plot_bgcolor="white",
    paper_bgcolor="white",
    font_family="Plus Jakarta Sans"
)

# Keterangan: Menggabungkan semua komponen HTML dan menutup container div
narasi_html = f"""
  <div class="narasi-box">
    Dengan <b>n = {n_sampel} responden</b> dari populasi N = {N_populasi:,}, diperoleh
    <b>Margin of Error (MoE) = ±{MoE*100:.2f}%</b> pada tingkat kepercayaan 95% setelah
    koreksi populasi terbatas (FPC). Nilai ini lebih kecil dari batas toleransi 10% dan
    melebihi sampel minimum Cochran ({n_cochran:.0f} responden), sehingga data dianggap
    <b>representatif</b> untuk menggambarkan populasi mahasiswa ITSB.
  </div>
</div>
"""

# Cetak struktur HTML bagian atas ke Colab
display(HTML(sampling_html))

# Cetak grafik Plotly 2D Interaktif di tengah komponen
fig.show()

# Cetak struktur HTML bagian bawah (Narasi) ke Colab
display(HTML(narasi_html))
```
